# Week 4: Advanced Project with Streamlit App

This notebook prepares the final Streamlit app for the Flickr8k image captioning project.

## 1. Save trained model artifacts

Run this cell after Week 3 training, when `improved_model`, `itos`, `stoi`, and `comparison_df` already exist in memory.

```python
import os, json, torch

os.makedirs("week4_streamlit_app/artifacts", exist_ok=True)

# Save improved model weights
# This expects the trained Week 3 model variable: improved_model
torch.save(improved_model.state_dict(), "week4_streamlit_app/artifacts/resnet50_lstm_captioner.pt")

# Save vocabulary
with open("week4_streamlit_app/artifacts/vocab.json", "w", encoding="utf-8") as f:
    json.dump({"itos": itos, "stoi": stoi}, f, ensure_ascii=False, indent=2)

# Save model config
config = {
    "encoder": "resnet50",
    "decoder": "lstm",
    "feature_dim": 2048,
    "embed_dim": 256,
    "hidden_dim": 512,
    "dropout": 0.3,
    "max_len": 25
}

with open("week4_streamlit_app/artifacts/model_config.json", "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2)

print("Artifacts saved successfully!")
print("week4_streamlit_app/artifacts/resnet50_lstm_captioner.pt")
print("week4_streamlit_app/artifacts/vocab.json")
print("week4_streamlit_app/artifacts/model_config.json")
```
